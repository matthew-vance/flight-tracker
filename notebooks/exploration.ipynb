# ADS-B Data Exploration

```python
import duckdb

PARQUET_GLOB = '../consumer/data/**/*.parquet'

def q(sql: str):
    return duckdb.sql(sql).df()
```

## Sanity check

```python
q(f"""
SELECT
    COUNT(*) AS total_messages,
    MIN(kafka_timestamp) AS earliest,
    MAX(kafka_timestamp) AS latest
FROM read_parquet('{PARQUET_GLOB}', hive_partitioning=true)
""")
```

## Most-seen aircraft

```python
q(f"""
SELECT
    hex_ident,
    COUNT(*) AS messages
FROM read_parquet('{PARQUET_GLOB}', hive_partitioning=true)
WHERE hex_ident IS NOT NULL AND hex_ident != ''
GROUP BY hex_ident
ORDER BY messages DESC
LIMIT 20
""")
```
